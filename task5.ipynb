```python
###############################################################
#     FULL EDA NOTEBOOK — gender_submission.csv (Windows Path)
###############################################################

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")

# =======================
# 1. Load Dataset
# =======================
df = pd.read_csv(
    r"C:\Users\9a33n\OneDrive\Documents\gender_submission.csv"
)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

display(df.head())

# =======================
# 2. Basic Info
# =======================
print("----- DATA INFO -----")
df.info()

print("\n----- STATISTICS -----")
display(df.describe(include="all"))

# =======================
# 3. Missing Values
# =======================
print("----- MISSING VALUES -----")
print(df.isnull().sum())

# =======================
# 4. Unique Values Per Column
# =======================
print("\n----- UNIQUE VALUES -----")
for col in df.columns:
    print(f"{col}: {df[col].nunique()}")

# =======================
# 5. Survived Column Analysis
# =======================
print("\n----- SURVIVED VALUE COUNTS -----")
print(df["Survived"].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)
plt.title("Survived Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.grid(axis='y', alpha=0.3)
plt.show()

# =======================
# 6. PassengerId Distribution
# =======================
plt.figure(figsize=(7,4))
plt.hist(df["PassengerId"], bins=20)
plt.title("PassengerId Distribution")
plt.xlabel("PassengerId")
plt.ylabel("Frequency")
plt.show()

# =======================
# 7. Correlation Heatmap
# =======================
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# =======================
# 8. Pairplot
# =======================
sns.pairplot(df)
plt.suptitle("Pairplot", y=1.02)
plt.show()

# =======================
# 9. Summary Insights
# =======================
print("\n----- SUMMARY INSIGHTS -----")
print("""
1. gender_submission.csv contains only PassengerId and Survived.
2. No missing values exist.
3. Survived distribution shows predicted survival outcomes.
4. Limited EDA possible (only 2 numeric columns).
5. For full Titanic EDA, use train.csv.
""")

# =======================
# 10. Save Cleaned File (Same Folder)
# =======================
save_path = r"C:\Users\9a33n\OneDrive\Documents\clean_gender_submission.csv"
df.to_csv(save_path, index=False)

print(f"Clean dataset saved successfully at:\n{save_path}")
```
